```python
data = [2, 3, 4, 2, 5, 2, 3, 6, 5]

frequency = {}

for number in data:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1

print("Frequency of each element:", frequency)
```

```python
def calculate_result(marks):
    total = sum(marks)
    percentage = (total / 500) * 100

    if percentage >= 90:
        grade = "A"
    elif percentage >= 75:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 40:
        grade = "D"
    else:
        grade = "F"

    return total, percentage, grade


# Marks in five subjects
marks = [85, 90, 78, 88, 95]

total, percentage, grade = calculate_result(marks)

print("Total Marks:", total)
print("Percentage:", percentage, "%")
print("Grade:", grade)
```

```python
employees = {
    "Amit": 45000,
    "Neha": 55000,
    "Riya": 70000,
    "Raj": 65000
}

# Increase every salary by 10%
for name in employees:
    employees[name] = employees[name] * 1.10

print("Updated Salaries:", employees)

# Find the highest-paid employee
highest_paid = max(employees, key=employees.get)

print("Highest-Paid Employee:", highest_paid)
print("Highest Salary:", employees[highest_paid])

# Calculate the average salary
average_salary = sum(employees.values()) / len(employees)

print("Average Salary:", average_salary)
```

```python

students = {
    "Rahul": {"Math": 90, "Science": 88},
    "Aman": {"Math": 78, "Science": 80},
    "Neha": {"Math": 95, "Science": 92}
}

for name, marks in students.items():
    total = sum(marks.values())
    average = total / len(marks)
    top_score = max(marks.values())

    print("Student:", name)
    print("Total Marks:", total)
    print("Average Marks:", average)
    print("Top Score:", top_score)
    print("--------------------")
```

```python

balance = 1000

# Deposit function
def deposit():
    global balance
    try:
        amount = float(input("Enter deposit amount: "))
        if amount <= 0:
            print("Amount must be greater than zero.")
        else:
            balance += amount
            print("Deposit successful!")
    except ValueError:
        print("Invalid amount. Please enter a number.")

# Withdraw function
def withdraw():
    global balance
    try:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= 0:
            print("Amount must be greater than zero.")
        elif amount > balance:
            raise ValueError("Insufficient balance!")
        else:
            balance -= amount
            print("Withdrawal successful!")
    except ValueError as e:
        print(e)

# Check balance function
def check_balance():
    print("Current balance:", balance)

# Menu
while True:
    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        deposit()
    elif choice == "2":
        withdraw()
    elif choice == "3":
        check_balance()
    elif choice == "4":
        print("Thank you!")
        break
    else:
        print("Invalid choice. Please try again.")
```
